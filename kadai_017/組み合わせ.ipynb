```python
class Human:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def check_adult(self):
       if self.age > 19:
          print(f"名前は{self.name}、成人です。")
       else:
           print(f"名前は{self.name}、未成年です。")

taro = Human("侍太郎", 36)
hanako = Human("侍花子", 30)
ichiro = Human("侍一郎", 5)
jiro = Human("侍次郎", 3)

human_list = [taro, hanako, ichiro, jiro]

for human in human_list:
    human.check_adult()
```
